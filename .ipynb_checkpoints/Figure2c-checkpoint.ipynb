```python
bias = ((mod[1:]-mod[0])).sel(latitude=slice(-15,0),longitude=slice(240,275)).mean(('latitude','longitude'))
stddev = np.sqrt(((std[1:]-std[0])**2).sel(latitude=slice(-15,0),longitude=slice(240,275)).mean(('latitude','longitude')))

# plot
fig, ax = plt.subplots(figsize=(6, 6))
sc = ax.scatter(bias, stddev, s=80, c='blue')

color = ["blue","orange","lime","green","deepskyblue","coral","purple","pink","navy"]
models = ['UFS P8','CMCC CM2','ECMWF IFS','HadGEM3 GC31','INM CM5','MPI ESM1','GFDL CM4','HiRAM SIT','FGOALS f3']

sc = ax.scatter(0.0, 0.0, s=80, c='k',label='OISSTv2')
for i, model in enumerate(models):
    sc = ax.scatter(bias[i], stddev[i], s=80, c=color[i],label=models[i])
    ax.text(bias[i] + 0.03, stddev[i], model, fontsize=8)

    

ax.text(0.03, 0.01, 'OISSTv2', fontsize=8)
ax.set_xlabel('Mean Bias')
ax.set_ylabel('RMSE of Standard Deviation')
#ax.set_title('Modified Taylor-like Plot')
ax.grid(linestyle='--')
ax.set_ylim(0,0.4)
ax.set_xlim(-0.8,1.8)
ax.legend()
plt.show()
```

```python
import numpy as np
import xarray as xr
import dask.array as da
from scipy.stats import linregress

def reg(index: xr.DataArray, data: xr.DataArray) -> tuple[xr.DataArray, xr.DataArray]:
    """
    Dask를 활용하여 3D 데이터 (data)와 1D 인덱스 (index) 간 선형 회귀 분석을 병렬 처리하는 함수.

    매개변수:
    - index: xarray DataArray of shape (model,), 회귀분석의 독립변수
    - data: xarray DataArray of shape (model, latitude, longitude), 분석할 데이터

    반환값:
    - (latitude, longitude) 차원의 xarray DataArray, 회귀계수(기울기)와 p-value
    """

    # 인덱스와 데이터 값을 Dask 배열로 변환
    X = da.from_array(index.values, chunks=(index.sizes['time'],))
    Y = da.from_array(data.values, chunks=(data.sizes['model'], data.sizes['latitude'], data.sizes['longitude']))

    # 회귀 분석 수행 (Dask map_blocks 사용)
    def lin_regression(y, X):
        """ 단일 (lat, lon) 위치에서 선형 회귀 수행 및 p-value 계산 """
        if np.all(np.isnan(y)):  # 모든 값이 NaN이면 NaN 반환
            return np.nan, np.nan
        valid_mask = np.isfinite(y)  # NaN이 없는 값만 선택
        if np.sum(valid_mask) < 2:  # 최소 2개 이상의 유효한 데이터 필요
            return np.nan, np.nan
        
        # 회귀분석 수행
        slope, intercept, r_value, p_value, std_err = linregress(X[valid_mask], y[valid_mask])
        
        return slope, p_value  # 기울기와 p-value 반환

    # 병렬 연산 수행
    regression_results = da.apply_along_axis(lin_regression, 0, Y, X)

    # 결과 분리 (slope, p-value)
    slope_result = regression_results[0].compute()
    p_value_result = regression_results[1].compute()

    # 결과를 xarray DataArray로 변환
    regression = xr.DataArray(
        slope_result,
        dims=['latitude', 'longitude'],
        coords={'latitude': data.coords['latitude'],
                'longitude': data.coords['longitude']},
        name="Regression"
    )

    p_value = xr.DataArray(
        p_value_result,
        dims=['latitude', 'longitude'],
        coords={'latitude': data.coords['latitude'],
                'longitude': data.coords['longitude']},
        name="P-value"
    )

    return regression, p_value
```

```python
sst = sst_regres.rename("sst")

import xarray as xr
import numpy as np
import xesmf as xe

# 1. 원시 SST DataArray
# sst: (lat, lon) 형태의 DataArray

# 2. 타겟 그리드 생성 (0.25도 간격)
target_grid = xr.Dataset({
    'lat': (['lat'], np.arange(-90, 90.01, 0.25)),
    'lon': (['lon'], np.arange(0, 360.01, 0.25))
})

# 3. 마스크 생성 (해수면만 True, 육지는 NaN 또는 False)
mask = xr.where(sst.notnull(), 1.0, np.nan)

# 4. Regridder 객체 생성
regridder_data = xe.Regridder(sst, target_grid, 'bilinear', periodic=True, reuse_weights=False)
regridder_mask = xe.Regridder(mask, target_grid, 'nearest_s2d', periodic=True, reuse_weights=False)

# 5. SST와 마스크 모두 리그리딩
sst_interp = regridder_data(sst)
mask_interp = regridder_mask(mask)

# 6. 마스크를 적용하여 육지에 NaN 유지
sst_interp_masked = sst_interp.where(mask_interp > 0.5)

# 7. 결과
sst_interp_masked.name = 'sst'

print(sst_interp_masked)
```
