```python
# Parameters
B = 1000   # Number of bootstrap samples
N = 24    # Number of spatial points per sample

# Step 1: Time average
tp_mean = tp.mean('time')        # (scenario, lat, lon)
conv_mean = conv.mean('time')    # (scenario, lat, lon)

# Step 2: Spatial average for main values
mpr = tp_mean.mean(('latitude', 'longitude'))         # (scenario,)
mconv = conv_mean.mean(('latitude', 'longitude'))     # (scenario,)
mlg = mpr - mconv                                      # large-scale

# Step 3: Compute relative change against CTL (index 0)
rpr = (mpr[1:] - mpr[0]) / mpr[0] * 100
rco = (mconv[1:] - mconv[0]) / mconv[0] * 100
rlg = (mlg[1:] - mlg[0]) / mlg[0] * 100

# Step 4: Bootstrap preparation
tp_flat = tp_mean.stack(points=('latitude', 'longitude'))      # (scenario, points)
conv_flat = conv_mean.stack(points=('latitude', 'longitude'))

rpr_boot = np.zeros((B, tp_flat.shape[0] - 1))
rco_boot = np.zeros((B, conv_flat.shape[0] - 1))
rlg_boot = np.zeros((B, conv_flat.shape[0] - 1))

# Step 5: Bootstrap loop
for b in range(B):
    idx = np.random.choice(tp_flat.points.size, N, replace=True)

    pr_sample = tp_flat[:, idx]         # (scenario, N)
    conv_sample = conv_flat[:, idx]

    ctl_pr = pr_sample[0]
    ctl_conv = conv_sample[0]

    pr_mean = pr_sample.mean(axis=1)
    conv_mean_sample = conv_sample.mean(axis=1)
    larg_mean = pr_mean - conv_mean_sample

    rpr_boot[b, :] = (pr_mean[1:] - pr_mean[0]) / pr_mean[0] * 100
    rco_boot[b, :] = (conv_mean_sample[1:] - conv_mean_sample[0]) / conv_mean_sample[0] * 100
    rlg_boot[b, :] = (larg_mean[1:] - larg_mean[0]) / larg_mean[0] * 100

# Step 6: Percentile range (2.5th, 97.5th) for 95% confidence interval
rpr_p2 = np.percentile(rpr_boot, 2.5, axis=0)
rpr_p97 = np.percentile(rpr_boot, 97.5, axis=0)

rco_p2 = np.percentile(rco_boot, 2.5, axis=0)
rco_p97 = np.percentile(rco_boot, 97.5, axis=0)

rlg_p2 = np.percentile(rlg_boot, 2.5, axis=0)
rlg_p97 = np.percentile(rlg_boot, 97.5, axis=0)

# Step 7: Reorder for plotting
tot_yerr = [[tot_pr[i] - rpr_p2[new_order[i]] for i in range(6)],
            [rpr_p97[new_order[i]] - tot_pr[i] for i in range(6)]]

conv_yerr = [[conv_pr[i] - rco_p2[new_order[i]] for i in range(6)],
             [rco_p97[new_order[i]] - conv_pr[i] for i in range(6)]]

larg_yerr = [[larg_pr[i] - rlg_p2[new_order[i]] for i in range(6)],
             [rlg_p97[new_order[i]] - larg_pr[i] for i in range(6)]]


# Step 8: Plotting
xax = np.arange(1, 7)
xnam = ['+1.0K', '+0.5K', '-0.5K', '-1.0K', 'Wet', 'Dry']

fig = plt.figure(figsize=(6, 4))
plt.subplots_adjust(bottom=0.3, top=0.95, left=0.05, right=0.8)

plt.bar(xax - 0.3, tot_pr, width=0.2, label='Total Precipitation', color='gray', capsize=3)
plt.bar(xax, conv_pr, width=0.2, label='Convective precip', color='red', capsize=3)
plt.bar(xax + 0.3, larg_pr, width=0.2, label='Large-scale precip', color='blue', capsize=3)

plt.xticks(xax, xnam)
plt.axhline(y=0, color='black', linestyle='-', linewidth=1)
plt.ylim(-25, 20)

legend = plt.legend(loc='lower right')
legend.get_frame().set_visible(False)
plt.show()
```

```python
# Parameters
B = 10000   # Number of bootstrap samples
N = 24    # Number of spatial points per sample

# Step 1: Time average
tp_mean = tp.mean('time')        # (scenario, lat, lon)
conv_mean = conv.mean('time')    # (scenario, lat, lon)

# Step 2: Spatial average for main values
mpr = tp_mean.mean(('latitude', 'longitude'))         # (scenario,)
mconv = conv_mean.mean(('latitude', 'longitude'))     # (scenario,)
mlg = mpr - mconv                                      # large-scale

# Step 3: Compute relative change against CTL (index 0)
rpr = (mpr[1:] - mpr[0]) / mpr[0] * 100
rco = (mconv[1:] - mconv[0]) / mconv[0] * 100
rlg = (mlg[1:] - mlg[0]) / mlg[0] * 100

# Step 4: Bootstrap preparation
tp_flat = tp_mean.stack(points=('latitude', 'longitude'))      # (scenario, points)
conv_flat = conv_mean.stack(points=('latitude', 'longitude'))

rpr_boot = np.zeros((B, tp_flat.shape[0] - 1))
rco_boot = np.zeros((B, conv_flat.shape[0] - 1))
rlg_boot = np.zeros((B, conv_flat.shape[0] - 1))

# Step 5: Bootstrap loop
for b in range(B):
    idx = np.random.choice(tp_flat.points.size, N, replace=True)

    pr_sample = tp_flat[:, idx]         # (scenario, N)
    conv_sample = conv_flat[:, idx]

    ctl_pr = pr_sample[0]
    ctl_conv = conv_sample[0]

    pr_mean = pr_sample.mean(axis=1)
    conv_mean_sample = conv_sample.mean(axis=1)
    larg_mean = pr_mean - conv_mean_sample

    rpr_boot[b, :] = (pr_mean[1:] - pr_mean[0]) / pr_mean[0] * 100
    rco_boot[b, :] = (conv_mean_sample[1:] - conv_mean_sample[0]) / conv_mean_sample[0] * 100
    rlg_boot[b, :] = (larg_mean[1:] - larg_mean[0]) / larg_mean[0] * 100

# Reorder bootstrap arrays to match plotting order
new_order = [0, 1, 3, 2, 4, 5]
xnam = ['+1.0K', '+0.5K', '-0.5K', '-1.0K', 'Wet', 'Dry']
xpos = np.arange(1, 7)

# Offset for each precipitation type
offset = 0.2

# Reordered data
rpr_reordered = [rpr_boot[:, i] for i in new_order]
rco_reordered = [rco_boot[:, i] for i in new_order]
rlg_reordered = [rlg_boot[:, i] for i in new_order]

fig, ax = plt.subplots(figsize=(10, 5))

# Total Precipitation (left side)
bp1 = ax.boxplot(rpr_reordered,
                 positions=xpos - offset,
                 patch_artist=True,
                 widths=0.18,
                 boxprops=dict(facecolor='gray'),
                 medianprops=dict(color='black'),
                 showfliers=False)

# Convective Precipitation (center)
bp2 = ax.boxplot(rco_reordered,
                 positions=xpos,
                 patch_artist=True,
                 widths=0.18,
                 boxprops=dict(facecolor='red'),
                 medianprops=dict(color='black'),
                 showfliers=False)

# Large-scale Precipitation (right side)
bp3 = ax.boxplot(rlg_reordered,
                 positions=xpos + offset,
                 patch_artist=True,
                 widths=0.18,
                 boxprops=dict(facecolor='blue'),
                 medianprops=dict(color='black'),
                 showfliers=False)

# Labels and layout
ax.set_xticks(xpos)
ax.set_xticklabels(xnam)
ax.axhline(y=0, color='black', linestyle='--', linewidth=1)
ax.set_ylabel('Relative Change (%)')
ax.set_title('Bootstrap-based Relative Precipitation Change (95% CI)')

# Custom legend
from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='gray', label='Total Precipitation'),
    Patch(facecolor='red', label='Convective Precipitation'),
    Patch(facecolor='blue', label='Large-scale Precipitation')
]
ax.legend(handles=legend_elements, loc='lower right')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
B = 10000
N = 24
new_order = [0, 1, 3, 2, 4, 5, 6]
xnam = ['W1.0', 'W0.5', 'C0.5', 'C1.0', 'Wet', 'Dry', 'BC']
xpos = np.arange(1, 8)

# Step 1: Time mean
tp_mean = tp.mean('time')         # shape: (scenario, lat, lon)
conv_mean = conv.mean('time')

# Step 2: Flatten
tp_flat = tp_mean.stack(points=('latitude', 'longitude'))       # (scenario, points)
conv_flat = conv_mean.stack(points=('latitude', 'longitude'))

# Step 3: Bootstrap storage (absolute only)
dpr_boot = np.zeros((B, tp_flat.shape[0] - 1))
dco_boot = np.zeros((B, conv_flat.shape[0] - 1))
dlg_boot = np.zeros((B, conv_flat.shape[0] - 1))

for b in range(B):
    idx = np.random.choice(tp_flat.points.size, N, replace=True)

    pr_sample = tp_flat[:, idx]
    conv_sample = conv_flat[:, idx]

    pr_mean = pr_sample.mean(axis=1)
    conv_mean_sample = conv_sample.mean(axis=1)
    larg_mean = pr_mean - conv_mean_sample

    dpr_boot[b, :] = pr_mean[1:] - pr_mean[0]
    dco_boot[b, :] = conv_mean_sample[1:] - conv_mean_sample[0]
    dlg_boot[b, :] = larg_mean[1:] - larg_mean[0]

# Step 4: Reorder for plotting
dpr_reordered = [dpr_boot[:, i] for i in new_order]
dco_reordered = [dco_boot[:, i] for i in new_order]
dlg_reordered = [dlg_boot[:, i] for i in new_order]

# Step 5: Plotting - 3 subplots
fig, axs = plt.subplots(1, 3, figsize=(15, 5), sharey=True)

# Total
axs[0].boxplot(dpr_reordered, patch_artist=True, widths=0.6,
               boxprops=dict(facecolor='gray'), medianprops=dict(color='black'),
               showfliers=False)
axs[0].set_title('(a) Total Precipitation',loc='left',fontsize=15)
axs[0].set_ylabel('Δ Precipitation',fontsize=12)
axs[0].set_xticks(xpos)
axs[0].set_xticklabels(xnam, rotation=45,fontsize=12)
axs[0].axhline(y=0, color='black', linestyle='--', linewidth=1)

# Convective
axs[1].boxplot(dco_reordered, patch_artist=True, widths=0.6,
               boxprops=dict(facecolor='red'), medianprops=dict(color='black'),
               showfliers=False)
axs[1].set_title('(b) Convective Precipitation',loc='left',fontsize=15)
axs[1].set_xticks(xpos)
axs[1].set_xticklabels(xnam, rotation=45,fontsize=12)
axs[1].axhline(y=0, color='black', linestyle='--', linewidth=1)

# Large-scale
axs[2].boxplot(dlg_reordered, patch_artist=True, widths=0.6,
               boxprops=dict(facecolor='blue'), medianprops=dict(color='black'),
               showfliers=False)
axs[2].set_title('(c) Large-scale Precipitation',loc='left',fontsize=15)
axs[2].set_xticks(xpos)
axs[2].set_xticklabels(xnam, rotation=45,fontsize=12)
axs[2].axhline(y=0, color='black', linestyle='--', linewidth=1)

plt.tight_layout()
plt.show()
```
