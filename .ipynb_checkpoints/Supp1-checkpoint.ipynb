```python
# ==============================================================
import numpy as np
import warnings

import cartopy.crs as ccrs
import cartopy.feature as cfeature
from cartopy.mpl.ticker import LongitudeFormatter, LatitudeFormatter
from cartopy.feature import ShapelyFeature
import cartopy.io.shapereader as shpreader


import sys
sys.path.append("/home/nchoi21/utils/python/nbchoi")
import f_nchoi as nb

import colormaps as cmaps
import matplotlib as mpl
import matplotlib.patches as patches
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.ticker import (MultipleLocator, AutoMinorLocator)
from scipy import signal

import os 
import glob
import subprocess
import regionmask

import pandas as pd
import xarray as xr
warnings.filterwarnings("ignore", category=UserWarning)
# ==============================================================
```

```python
## OPEN UFS FILES 
mod = xr.open_dataset('/scratch/nchoi21/CMIP/HighResMIP/climate/Regrid_25yrs_ann_clim_pr.nc').pr
print(mod.coords['model'])

## GPCP
mod[0] = xr.open_dataset('/scratch/nchoi21/GPCP/GPCP3.3/regrid/GPCP3.3_25yrs_ann_clim_pr.nc').pr.data

```

```python
zm = mod.mean('longitude')
#zm = zm - zm[0]
mme = zm[1:].mean('model')
#mme = zm[3]
#mme = zm[[2,5]].mean('model')
#mme = mme - zm[0]
cmip = zm[2:].mean('model')
print(mme)
```

```python
## PLOT Weekly Bias
fig = plt.figure(figsize=(8,4))
plt.subplots_adjust(bottom=0.2,top=0.95,left=0.05,right=0.8)
lat_tick = np.arange(-90,91,10)
lat_formatter = LatitudeFormatter()

plt.subplot(1,1,1,projection=ccrs.PlateCarree())
plt.subplots_adjust(bottom=0.2, top=0.95, left=0.05, right=0.8)
clevs = np.arange(1,13,1)
a=plt.contourf(mme.coords['time']*10,mme.coords['latitude'],mme.T,levels=clevs,extend='both')
plt.gca().set_yticks(lat_tick)
plt.gca().set_xticks(np.arange(10,130,10))
plt.gca().set_xticklabels(['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'])
plt.gca().yaxis.set_major_formatter(lat_formatter)
plt.gca().set_extent([10, 120, -30, 30], crs=ccrs.PlateCarree())
plt.gca().gridlines(color='lightgray', linestyle='--', linewidth=0.5, zorder=3)
plt.set_cmap(cmaps.WhiteBlueGreenYellowRed)
plt.gca().set_title('Monthly Precipitation Bias (MME)',loc='left',fontsize=10)
cax = plt.axes([0.075,0.1,0.7,0.03]) # Left, Bottom, width, height
cb = plt.colorbar(cax=cax,orientation='horizontal',ticks=clevs,extend='both')
```
